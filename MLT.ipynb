MEAN IMPUTATION

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame
df = pd.DataFrame([
    [3, 2, np.nan],
    [2, np.nan, 6],
    [np.nan, 8, 9]
])

# Impute missing values with the column mean
imputed_df = df.fillna(df.mean())

print(imputed_df)
```
